```python
#Read
print("B1. Kopi, kepahitan > 3:")
for m in menu.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}}, {"_id": 0}):
    print("  ", m)

print("\nB2. Harga 15000 - 30000:")
for m in menu.find({"harga": {"$gte": 15000, "$lte": 30000}}, {"_id": 0, "nama_menu": 1, "kategori": 1, "harga": 1}):
    print("  ", m)

print("\nB3. Tanpa field pedas_level:")
for m in menu.find({"pedas_level": {"$exists": False}}, {"_id": 0, "nama_menu": 1, "kategori": 1}):
    print("  ", m)
```

```python
#Update
r1 = menu.update_many({}, {"$set": {"status": "Tersedia"}})
print("Dokumen terubah:", r1.modified_count)

menu.update_one({"nama_menu": "Espresso"}, {"$set": {"status": "Habis"}})
menu.update_one({"nama_menu": "Ayam Geprek"}, {"$set": {"status": "Habis"}})

for m in menu.find({}, {"_id": 0, "nama_menu": 1, "status": 1}):
    print("  ", m)
```

```python
#Spark
from pyspark.sql import SparkSession
from pyspark.sql.types import (StructType, StructField, StringType,
                               IntegerType, BooleanType)
from pyspark.sql.functions import avg, count, round as sround

spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

skema = StructType([
    StructField("nama_menu", StringType(), True),
    StructField("kategori", StringType(), True),
    StructField("harga", IntegerType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("stok", IntegerType(), True),
    StructField("rekomendasi_barista", BooleanType(), True),
    StructField("status", StringType(), True),
])

dokumen = list(menu.find({}, {"_id": 0}))
baris = [tuple(doc.get(f.name) for f in skema.fields) for doc in dokumen]
df_menu = spark.createDataFrame(baris, schema=skema)
```

```python
#D1
df_menu.printSchema()
df_menu.show(truncate=False)
```

```python
#D2
df_menu.groupby("kategori").agg(
    sround(avg("harga"), 2).alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show()

spark.stop()
```
